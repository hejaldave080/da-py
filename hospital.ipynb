```python
#GROUPBY AND ANALYSIS
#Diagnosis: Find patient count, total billing and average billing for each Diagnosis
Diagnosis=df.groupby("Diagnosis").agg(PatientCount=("Patient_ID","count"),
                                      TotalBilling=("Total_Bill","sum"),
                                      AverageBilling=("Total_Bill","mean"),)
Diagnosis
```

```python
#City: Find patient count, total billing and average billing for each City.
City=df.groupby("City").agg(PatientCount=("Patient_ID","count"),
                            totalbilling=("Total_Bill","sum"),
                            averagebill=("Total_Bill","mean"))
City
```

```python
#Admission Type: Find patient count, average billing, minimum billing and maximum billing for each Admission_Type
AdmissionType=df.groupby("Admission_Type").agg(PatientCount=("Patient_ID","count"),
                                               averagebilling=("Total_Bill","mean"),
                                               minbill=("Total_Bill","min"),
                                               maxbill=("Total_Bill","max"))
AdmissionType
```

```python
#Insurance: Find total Insurance_Covered and total Patient_Paid for each Insurance_Type.
insurance=df.groupby("Insurance_Type").agg(insuranceCovered=("Insurance_Covered","sum"),
                                      patientpaid=("Patient_Paid","sum"),
                                      )
insurance
```

```python
# Multiple Aggregations using agg()
# Department: calculate patient count, total billing, average billing, minimum billing and maximum billing.
Department=df.groupby("Department").agg(PatientCount=("Patient_ID","count"),
                                       averagebilling=("Total_Bill","mean"),
                                     minbill=("Total_Bill","min"),
                                       maxbill=("Total_Bill","max"),
                                       TotalBill=("Total_Bill","sum"))
Department  
```
